```python
import pandas as pd
```

```python
data = {
    "Student_ID": [1, 2, 3, 4, 5],
    "Name": ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    "Attendance": [85, 90, 78, 92, 88],
    "Final_Grade": ['A', 'B', 'C', 'A', 'B'],
    "Enrolled_Date": ['2024-01-10', '2024-02-15', '2024-01-25', '2024-03-01', '2024-01-20']
}
df = pd.DataFrame(data)
df
```

```python
print("Column names: ", df.columns)
print("Column data types: \n", df.dtypes)
```

```python
df["Enrolled_Date"] = df["Enrolled_Date"].astype("datetime64[s]")
df["Enrolled_Date"]
```

```python
df.dtypes
```

```python
df["Final_Grade"].unique()
```

```python
df["Final_Grade"] = df["Final_Grade"].astype("category")
```

```python
df["Final_Grade_Encode"] = df["Final_Grade"].cat.codes + 1
df
```

```python
Final_Grade_One_Hot_Encoding = pd.get_dummies(df["Final_Grade"])
Final_Grade_One_Hot_Encoding = Final_Grade_One_Hot_Encoding.astype(int)
df = pd.concat([df, Final_Grade_One_Hot_Encoding], axis=1)
df
```

```python
Enrolled_Year = df["Enrolled_Date"].dt.year
Enrolled_Year
```
